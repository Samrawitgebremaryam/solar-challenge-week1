```python
benin = pd.read_csv('../data/benin_clean.csv')
sierra_leone = pd.read_csv('../data/sierra_leone_clean.csv')
togo = pd.read_csv('../data/togo_clean.csv')
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
sns.boxplot(data=[benin['GHI'], sierra_leone['GHI'], togo['GHI']], palette='Set2')
plt.xticks([0, 1, 2], ['Benin', 'Sierra Leone', 'Togo'])
plt.title('GHI Comparison')
# Repeat for DNI, DHI
plt.tight_layout()
plt.show()
```

```python
summary = pd.DataFrame({
    'Country': ['Benin', 'Sierra Leone', 'Togo'],
    'GHI_Mean': [benin['GHI'].mean(), sierra_leone['GHI'].mean(), togo['GHI'].mean()],
    'GHI_Median': [benin['GHI'].median(), sierra_leone['GHI'].median(), togo['GHI'].median()],
    'GHI_SD': [benin['GHI'].std(), sierra_leone['GHI'].std(), togo['GHI'].std()]
    # Add DNI, DHI
})
print(summary)
```

```python
from scipy.stats import f_oneway
stat, p_value = f_oneway(benin['GHI'], sierra_leone['GHI'], togo['GHI'])
print(f"ANOVA p-value: {p_value}")
```

```python
### Key Observations
# - Benin has the highest median GHI but also the greatest variability.
# - Sierra Leone shows consistent DNI across months.
# - Togo has lower average DHI, possibly due to cloud cover.
```

```python
avg_ghi = pd.DataFrame({
    'Country': ['Benin', 'Sierra Leone', 'Togo'],
    'Avg_GHI': [benin['GHI'].mean(), sierra_leone['GHI'].mean(), togo['GHI'].mean()]
})
sns.barplot(x='Country', y='Avg_GHI', data=avg_ghi)
plt.title('Average GHI by Country')
plt.show()
```
